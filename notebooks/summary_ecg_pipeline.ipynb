# ECG Signal Processing — Summary Notebook

This notebook summarizes the main pipeline components in this repository and provides a small, self-contained, synthetic example demonstrating filtering, R-peak detection, BPM calculation, and QRS analysis.

## Key modules
- `ecg_parser`: loading records and annotations (MIT-BIH/WFDB).
- `filters`: bandpass filtering helper.
- `rpeak_detection`: R-peak detection with NeuroKit2 (optional) and SciPy fallback.
- `bpm_analysis`: RR interval and BPM calculations.
- `qrs_analysis`: QRS duration and amplitude estimates.
- `visualization`: save ECG plots with R-peaks marked.
